# Customer Segmentation using K-Means Clustering

## Project: Retail Store Customer Segmentation

This notebook uses K-Means clustering to group customers based on **Annual Income** and **Spending Score**.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

## 2. Load Dataset

```python
from google.colab import files

uploaded = files.upload()
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

df.head()
```

## 3. Check Dataset Information

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

df.info()
```

## 4. Check Missing Values

```python
df.isnull().sum()
```

## 5. Select Features for Clustering

```python
X = df[["Annual Income (k$)", "Spending Score (1-100)" ]].copy()
X.head()
```

## 6. Visualize Customer Data

```python
plt.figure(figsize=(8, 5))
plt.scatter(X["Annual Income (k$)"], X["Spending Score (1-100)"])
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("Customers Before Clustering")
plt.show()
```

## 7. Find the Best Number of Clusters — Elbow Method

```python
wcss = []

for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=42, n_init=10)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker="o")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

### Observation

The elbow appears around **5 clusters**, so we use 5 customer groups.

## 8. Apply K-Means Clustering

```python
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
X["Cluster"] = kmeans.fit_predict(X)
X.head()
```

## 9. Visualize the 5 Customer Clusters

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    X["Annual Income (k$)"],
    X["Spending Score (1-100)"],
    c=X["Cluster"]
)
plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    s=200,
    marker="X"
)
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("Customer Segmentation using K-Means")
plt.show()
```

## 10. Number of Customers in Each Cluster

```python
X["Cluster"].value_counts().sort_index()
```

## 11. Average Income and Spending Score by Cluster

```python
cluster_summary = X.groupby("Cluster").mean()
cluster_summary
```

## 12. Interpret the Customer Segments

- **Cluster 0:** Average income + average spending
- **Cluster 1:** High income + high spending — high-value customers
- **Cluster 2:** Low income + high spending — high-spending customers with lower income
- **Cluster 3:** High income + low spending — customers with potential for targeted offers
- **Cluster 4:** Low income + low spending — low-value customers

## 13. Add Cluster Information to the Original Dataset

```python
df["Cluster"] = X["Cluster"]
df.head()
```

## 14. Save the Final Customer Segmentation Dataset

```python
df.to_csv("customer_segments.csv", index=False)
print("Saved as customer_segments.csv")
```

## Conclusion

K-Means clustering successfully divided the retail customers into 5 groups based on annual income and spending score. These segments can help a retail store understand customer behavior and create more targeted marketing strategies.
